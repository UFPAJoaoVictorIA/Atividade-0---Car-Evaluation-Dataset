```python
!pip install ucimlrepo
from ucimlrepo import fetch_ucirepo

# fetch dataset
car_evaluation = fetch_ucirepo(id=19)

# data (as pandas dataframes)
X = car_evaluation.data.features
y = car_evaluation.data.targets

# metadata
print(car_evaluation.metadata)

# variable information
print(car_evaluation.variables)

# Combine X and y into a single DataFrame 'df' for easier manipulation, if needed.
df = pd.concat([X, y], axis=1)
df.head()
```

```python
#conferir estrutura e tipos
print("Linhas e colunas:", df.shape)
df.info()
```

```python
#conferir valores ausentes e classes antes da conversão
print("Valores ausentes por coluna:")
print(df.isna().sum())
print("\nRegistros por classe:")
print(df["class"].value_counts())
```

No momento de agora , eu troco o valor selecionado para outro com fator importante para resultar em um resultado com valor e resposta correto.

```python
# Converter a classe para número para o dataset Car Evaluation
df['class'] = df['class'].map({
    'unacc': 0,
    'acc': 1,
    'good': 1,
    'vgood': 1
})
df.head()
```

11-Table2-1.png

```python
# Codificar as variáveis categóricas usando one-hot encoding
X_encoded = pd.get_dummies(df.drop('class', axis=1))

# separar as features (X) e a classe (y) em arrays NumPy
X = np.array(X_encoded, dtype=float)
y = np.array(df["class"], dtype=int)

print("X:", X.shape, "— linhas e atributos")
print("y:", y.shape, "— classes")
print("Primeira linha de features (X[0]):", X[0])
print("Classe do primeiro item (y[0]):", y[0])
```
